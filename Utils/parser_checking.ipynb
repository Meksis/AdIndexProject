```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

# Создайте экземпляр браузера
driver = webdriver.Chrome()

# Используйте execute_script для перехода на страницу асинхронно
driver.execute_script("window.location.href='https://adindex.ru/news/all.phtml';")
print('Сразу же идем дальше, не дожидаясь открытия страницы')

# Определите локатор для нужного элемента
element_locator = (By.ID, 'specific_element_id')

# Используйте WebDriverWait для ожидания появления элемента
try:
    element = WebDriverWait(driver, 10).until(
        EC.presence_of_element_located(element_locator)
    )
    # Элемент найден, можно работать с ним
    print("Элемент найден, выполняются действия...")
finally:
    # Закрыть вкладку
    driver.close()

# Закрыть браузер (опционально, если вы больше не собираетесь использовать браузер)
driver.quit()
```
