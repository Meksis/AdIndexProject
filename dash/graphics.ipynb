```python
grouped_author = df.groupby('Автор').agg(
    total_readers=pd.NamedAgg(column='Читатели', aggfunc='sum'),
    total_posts=pd.NamedAgg(column='post_tag', aggfunc='count'),
    avg_depth=pd.NamedAgg(column='Глубина просмотра', aggfunc='mean'),
    best_topic=pd.NamedAgg(column='post_tag', aggfunc=lambda x: df.loc[x.index, 'Читатели'].idxmax())).reset_index()

grouped_author['best_topic'] = grouped_author['best_topic'].apply(lambda x: df.loc[x, 'post_tag'])

grouped_author.columns = ['Автор', 'Количество просмотров', 'Количество статей', 'Глубина', 'Лучшая тема (По просмотрам)']

print(grouped_author)
```

```python
# Группировка данных по автору и теме, а затем рассчет среднего количества просмотров
grouped_data = pd.DataFrame(df.groupby(['Автор', 'post_tag'])['Читатели'].mean())

# Переименование столбца среднего количества просмотров для ясности
grouped_data.rename(columns={'Читатели': 'Среднее количество просмотров'}, inplace=True)

# Вывод первых нескольких строк таблицы
grouped_data
```

```python
average_views_per_author = df.groupby('author')['Читатели'].mean()
top_10_authors = average_views_per_author.sort_values(ascending=False).head(10)

print(top_10_authors)
```

```python
# Отсортируйте таблицу с средними просмотрами по убыванию
sorted_data = grouped_data.sort_values(by='Среднее количество просмотров', ascending=False)

# Выберите топ-10 авторов
top_10_authors = sorted_data.index.get_level_values(0).unique()[:10]

# Создаем список для хранения данных
data_list = []

# Для каждого автора из топ-10 выбираем темы, в которых он был наиболее популярен
for author in top_10_authors:
    author_data = sorted_data.loc[author]
    top_topic = author_data.idxmax()[0]  # Находим тему с максимальным средним количеством просмотров
    top_views = author_data.max()  # Получаем максимальное среднее количество просмотров
    
    # Получаем общее количество статей по этой теме для данного автора
    total_articles = df[(df['author'] == author) & (df['post_tag'] == top_topic)].shape[0]
    
    # Получаем общее количество просмотров статей данного authorа по этой теме
    total_views = df[(df['author'] == author) & (df['post_tag'] == top_topic)]['Читатели'].sum()
    
    # Добавляем данные в список
    data_list.append({'author': author, 
                      'post_tag': top_topic, 
                      'Среднее количество просмотров': top_views[0],
                      'Общее количество статей по теме': total_articles,
                      'Общее количество просмотров статей': total_views})

# Создаем DataFrame из списка
top_authors_by_topic = pd.DataFrame(data_list)

# Выводим результаты
top_authors_by_topic
```
