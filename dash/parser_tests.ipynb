```python
from Utils import parser

from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options

import pandas as pd
import datetime as dt
import os
```

```python
today = dt.date.today()
```

```python
result = parser.parse_some_news(dt.date(today.year, today.month, 1), today, next_page_delay=0.5)
```

```python
os.mkdir('csvs') if not os.path.exists('csvs') else ...

result.to_csv(f'csvs/AdIndex_parsed_{dt.date(today.year, today.month, 1)}-_-{today}.csv')
```

```python
parser.get_data_from_url(result.loc[0]['link'], dt.date(today.year, today.month, 1), today)
```

```python
metrics = pd.DataFrame()

for row_number in range(len(result)):
    link = result.loc[row_number]['link']

    metrics = pd.concat([metrics, pd.DataFrame.from_dict([parser.get_data_from_url(link, dt.date(today.year, today.month, 1), today)[-1]])])
```

```python
end = result.copy()
end = end.merge(metrics, on='link')
end
end.to_csv(f'csvs/AdIndex_Metrics_{dt.date(today.year, today.month, 1)}-_-{today}.csv')
```
