```python
cursor.execute('''CREATE TABLE IF NOT EXISTS RAW_DATA (
                post_id INTEGER PRIMARY KEY,
                date DATETIME NOT NULL,
                author TEXT NOT NULL,
                post_name TEXT NOT NULL,
                post_tag TEXT NOT NULL,
                link TEXT NOT NULL
            );''')

con.commit()

cursor.execute('''CREATE TABLE IF NOT EXISTS METRICS_DATA (
                date DATETIME NOT NULL,
                author TEXT NOT NULL,
                post_name TEXT NOT NULL,
                post_tag TEXT NOT NULL,
                post_id INTEGER PRIMARY KEY,
                link TEXT NOT NULL,
                visitors REAL NOT NULL,
                readers REAL NOT NULL,
                unique_users REAL NOT NULL,
                non_reading_rate REAL NOT NULL,
                view_depth REAL NOT NULL,
                avg_time_on_site REAL NOT NULL
            );''')

con.commit()
```
