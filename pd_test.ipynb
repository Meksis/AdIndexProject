```python
import pandas as pd
```

```python
df = pd.read_csv('dash\csvs\AdIndex main news METRICS 2023-12-14.csv')
df.head()
```

```python
import plotly.express as px
```

```python
sum_read = df.groupby('Автор')['Читатели'].sum().reset_index()
sum_read.sort_values(by='Читатели', ascending=False, inplace=True)
best_topic_author = px.bar(sum_read, x='Автор', y='Читатели', title='Рейтинг публикаций', 
            width=1400, height=800)


best_topic_author.show()
```
